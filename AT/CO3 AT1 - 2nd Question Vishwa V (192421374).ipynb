```python
import pandas as pd
import numpy as np

# Load dataset
data = pd.read_csv("/marks.csv")

x = data["Marks"]

# MLE Estimation
mle_mean = np.mean(x)
mle_variance = np.var(x)

print("MLE")
print("Mean =", mle_mean)
print("Variance =", mle_variance)

# Bayesian Estimation

prior_mean = 65
prior_weight = 5

posterior_mean = (prior_weight * prior_mean + len(x) * mle_mean) / (prior_weight + len(x))

print("\nBayesian Estimation")
print("Prior Mean =", prior_mean)
print("Posterior Mean =", posterior_mean)
```
